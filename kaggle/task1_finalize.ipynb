```python
import glob, os, shutil
os.environ["ATML_OUT"] = "/kaggle/working/task1_outputs"
os.environ["ATML_RESULTS"] = "/kaggle/working/task1_results"
for name in ("task1_outputs", "task1_results"):
    for previous in sorted(glob.glob(f"/kaggle/input/**/{name}", recursive=True)):
        print("restoring", previous)
        shutil.copytree(previous, f"/kaggle/working/{name}", dirs_exist_ok=True)
os.chdir("/kaggle/working/atml-pa3")
```

```python
!python -m task1.diagnostics
```

```python
import zipfile
from pathlib import Path
out, res = Path("/kaggle/working/task1_outputs"), Path("/kaggle/working/task1_results")
with zipfile.ZipFile("/kaggle/working/task1_package.zip", "w", zipfile.ZIP_DEFLATED) as package:
    for path in sorted(res.rglob("*")):
        if path.is_file():
            package.write(path, Path("task1_results") / path.relative_to(res))
    for pattern in ("logs/*.log", "queue_status.json", "runs/*/train_summary.json", "runs/*/train_log.jsonl",
                    "runs/*/*/train_log.jsonl"):
        for path in sorted(out.glob(pattern)):
            package.write(path, Path("task1_outputs") / path.relative_to(out))
    names = package.namelist()
print(len(names), "files")
shutil.rmtree(out)
shutil.rmtree(res)
shutil.rmtree("/kaggle/working/atml-pa3")
!ls -la /kaggle/working
```
